# Part 2: Analysis (A)

All analysis cells are read-only (`q(...)` on `gold.*`). The only exception is the cell that saves a mart, labeled `# SAVE MART`. Each question gets a 2-3 sentence conclusion after the query.

## A1. How do revenue and orders change over time?
- **Tables:** `fact_order_items`, `fact_orders`, `dim_date`
- **Method:** monthly revenue (`SUM(price)`), order count, average order value (revenue / orders), month-over-month growth.
- **Output:** DAX measures (`Revenue`, `Orders`, `Avg Order Value`, `MoM %`); no mart needed.
- **Conclusion:** _write after running_

## A2. Which states bring the most revenue?
- **Tables:** `fact_order_items`, `dim_customers`
- **Method:** revenue, orders and average order value per customer state; share of total revenue.
- **Output:** DAX measures filtered by `dim_customers[state]`.
- **Conclusion:** _write after running_

## A3. How well does delivery perform?
- **Tables:** `fact_orders`, `dim_customers` (delivered orders only)
- **Method:** average `delivery_days`, `is_late` share, average `delay_days` by state and by month. Check whether late deliveries are concentrated in certain states or months (for example Nov 2017).
- **Output:** DAX `Late Delivery %`, `Avg Delivery Days`; columns already in `fact_orders`.
- **Conclusion:** _write after running_

## A4. How do customers behave? (RFM)
- **Tables:** `fact_orders`, `fact_order_items`, `dim_customers`
- **Method:** repeat-customer rate (customers with 2+ orders), then RFM scores and segments.
- **Output:** saved mart `gold.mart_customer_rfm`.
- **Conclusion:** _write after running_

## A5. How do customers pay?
- **Tables:** `fact_orders`
- **Method:** share of orders by `main_payment_type`; average order value and installments by payment type.
- **Output:** DAX measures by `main_payment_type`.
- **Conclusion:** _write after running_
