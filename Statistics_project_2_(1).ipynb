## Test statistical hypotheses

# 0.05 səviyyəsindən qat-qat kiçik olduğu üçün Null hipotez H_0 rədd edildi. Bu o deməkdir ki, iki tarifin orta gəlirləri arasındakı fərq təsadüfi deyil, statistiki cəhətdən kəskin və mühümdür. Ultimate tarifində gəlir sabit və yüksəkdir .Surf tarifində isə orta gəlir xeyli aşağıdır, lakin bəzi istifadəçilər paket limitlərini aşaraq əlavə ödənişlər etdikləri üçün gəlirdə səpələnmələr müşahidə olunur.


```python
from scipy import stats


surf_revenue = df_monthly[df_monthly['plan'] == 'surf']['total_revenue']
ultimate_revenue = df_monthly[df_monthly['plan'] == 'ultimate']['total_revenue']


t_stat, p_value = stats.ttest_ind(surf_revenue, ultimate_revenue, equal_var=False)

print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")


alpha = 0.05
if p_value < alpha:
    print("Null hipotez rədd edilir: Tariflərin orta gəlirləri arasında statistiki cəhətdən mühüm fərq VAR.")
else:
    print("Null hipotez qəbul edilir: Tariflərin orta gəlirləri arasında əhəmiyyətli fərq YOXDUR.")
```

# Test nəticəsində bu dəyər 0.05-dən qat-qat kiçik olduğu üçün Null hipotez H_0 rədd edildi. Bu o deməkdir ki, NY-NJ regionunun orta gəliri ilə digər regionların gəliri arasındakı fərq təsadüfi deyil, statistiki cəhətdən çox güclü, kəskin və real fərqdir. Mənfi T-statistiki -8.31 göstərir ki, NY-NJ bölgəsinin orta gəlir göstəriciləri digər regionlardan əhəmiyyətli dərəcədə fərqlənir. Şirkət üçün bu region (NY-NJ) xüsusi diqqət tələb edən əsas seqmentlərdən biri ola bilər və buradakı istifadəçi davranışları ayrıca araşdırılmalıdır.

```python
from scipy import stats

# 1. Əgər city sütunu df_monthly-də yoxdursa, users cədvəlindən birləşdiririk
if 'city' not in df_monthly.columns:
    df_monthly = df_monthly.merge(users[['user_id', 'city']], on='user_id', how='left')

# 2. city sütununda 'NY' və ya 'New York' keçənləri ayırırıq
nynj_revenue = df_monthly[df_monthly['city'].str.contains('NY|New York', case=False, na=False)]['total_revenue']
other_revenue = df_monthly[~df_monthly['city'].str.contains('NY|New York', case=False, na=False)]['total_revenue']

# 3. T-testini həyata keçiririk
t_stat, p_value = stats.ttest_ind(nynj_revenue, other_revenue, equal_var=False)

print(f"T-statistic: {t_stat}")
print(f"P-value: {p_value}")

# 4. Nəticəni yoxlayırıq
alpha = 0.05
if p_value < alpha:
    print("Null hipotez rədd edilir: NY-NJ bölgəsinin orta gəliri ilə digər regionlar arasında fərq VAR.")
else:
    print("Null hipotez qəbul edilir: Bölgələr arasında orta gəlir baxımından fərq YOXDUR.")
```
